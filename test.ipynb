```python
import xml.etree.ElementTree as ET

def print_xml_tree(element, indent=""):
    """Recursively print the XML tree structure."""
    print(f"{indent}{element.tag}")
    for child in element:
        print_xml_tree(child, indent + "  ")

# Example usage with a decoded ProPresenter file
decoded_file_path = "adawd.pro"

# Read the decoded file
with open(decoded_file_path, "r") as file:
    decoded_data = file.read()

# Parse the decoded data as XML
root = ET.fromstring(decoded_data)

# Print the XML tree structure
print("XML Tree Structure:")
print_xml_tree(root)
```
